```python
df = pd.read_csv('../data/raw/deliveries.csv')
os.makedirs('../charts', exist_ok=True)

sns.set_theme(style="whitegrid")
print("Ready!")
```

```python
#Chart 1: Delayed vs On Time
sns.countplot(x='is_delayed', data=df, palette=['#2ecc71', '#e74c3c'])
plt.title('Delayed vs On Time Deliveries')
plt.xlabel('0 = On Time     1 = Delayed')
plt.ylabel('Count')
plt.savefig('../charts/chart1_delivery_status.png')
plt.show()
```

```python
#2Weather vs Delay
sns.boxplot(x='is_delayed', y='weather_severity', data=df,palette=['#2ecc71', '#e74c3c'])
plt.title('Weather Severity vs Delay')
plt.xlabel('0 = On Time     1 = Delayed')
plt.ylabel('Weather Severity (1=Clear, 5=Blizzard)')
plt.savefig('../charts/chart2_weather_vs_delay.png')
plt.show()
```

```python
#3Season vs Delay
sns.countplot(x='season', hue='is_delayed', data=df,palette=['#2ecc71', '#e74c3c'],order=['winter','spring','summer','fall'])
plt.title('Delays by Season')
plt.xlabel('Season')
plt.ylabel('Count')
plt.legend(labels=['On Time', 'Delayed'])
plt.savefig('../charts/chart3_season_vs_delay.png')
plt.show()
```
