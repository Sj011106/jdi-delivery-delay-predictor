```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os


df = pd.read_csv('../data/raw/deliveries.csv')
print("Data loaded successfully!")
print(f"Shape: {df.shape}")
print(f"\nColumn names: {list(df.columns)}")
print(f"\nFirst 5 rows:")
df.head()
```

```python
print("BASIC DATASET INFORMATION")

print(f"\nTotal deliveries:  {len(df)}")
print(f"Total features:    {df.shape[1]}")
print(f"\nColumn data types:")
print(df.dtypes)
print(f"\nMissing values per column:")
print(df.isnull().sum())
print(f"\nBasic statistics:")
df.describe()
```

```python
df = pd.read_csv('../data/raw/deliveries.csv')
os.makedirs('../charts', exist_ok=True)

sns.set_theme(style="whitegrid")
print("Ready!")
```

```python
#Chart 1: Delayed vs On Time
sns.countplot(x='is_delayed', data=df, palette=['#2ecc71', '#e74c3c'])
plt.title('Delayed vs On Time Deliveries')
plt.xlabel('0 = On Time     1 = Delayed')
plt.ylabel('Count')
plt.savefig('../charts/chart1_delivery_status.png')
plt.show()
```

```python
#2Weather vs Delay
sns.boxplot(x='is_delayed', y='weather_severity', data=df,palette=['#2ecc71', '#e74c3c'])
plt.title('Weather Severity vs Delay')
plt.xlabel('0 = On Time     1 = Delayed')
plt.ylabel('Weather Severity (1=Clear, 5=Blizzard)')
plt.savefig('../charts/chart2_weather_vs_delay.png')
plt.show()
```

```python
#3Season vs Delay
sns.countplot(x='season', hue='is_delayed', data=df,palette=['#2ecc71', '#e74c3c'],order=['winter','spring','summer','fall'])
plt.title('Delays by Season')
plt.xlabel('Season')
plt.ylabel('Count')
plt.legend(labels=['On Time', 'Delayed'])
plt.savefig('../charts/chart3_season_vs_delay.png')
plt.show()
```

```python
#4Road Type vs Delay
sns.countplot(x='road_type', hue='is_delayed', data=df,palette=['#2ecc71', '#e74c3c'])
plt.title('Delays by Road Type')
plt.xlabel('Road Type')
plt.ylabel('Count')
plt.legend(labels=['On Time', 'Delayed'])
plt.savefig('../charts/chart4_roadtype_vs_delay.png')
plt.show()
```

```python
#5Correlation Heatmap
numerical_cols = ['distance_km', 'weather_severity','cargo_weight_kg', 'driver_experience_yrs','departure_hour', 'is_delayed']
sns.heatmap(df[numerical_cols].corr(), annot=True,fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.savefig('../charts/chart5_correlation_heatmap.png')
plt.show()
```

```python
#6: Distance vs Delay
sns.histplot(data=df, x='distance_km', hue='is_delayed',palette=['#2ecc71', '#e74c3c'], bins=30)
plt.title('Distance Distribution by Delay Status')
plt.xlabel('Distance (km)')
plt.ylabel('Count')
plt.legend(labels=['Delayed', 'On Time'])
plt.savefig('../charts/chart6_distance_vs_delay.png')
plt.show()
```

```python
print("KEY FINDINGS")
print("=" * 40)
print(f"\nDelay rate: {df['is_delayed'].mean():.1%}")
print(f"\nWeather average:")
print(df.groupby('is_delayed')['weather_severity'].mean())
print(f"\nDelay rate by season:")
print(df.groupby('season')['is_delayed'].mean().sort_values(ascending=False))
print(f"\nDelay rate by road type:")
print(df.groupby('road_type')['is_delayed'].mean())
print(f"\nAll charts saved to charts/ folder!")
```
